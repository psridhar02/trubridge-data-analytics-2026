```python
import numpy as np
import pandas as pd

df = pd.read_csv('/content/Sept26_FINAL_sridhar_analysis_ready_dataset.csv')

# Stats & Risk Tiers
df['Food_Insecurity_Tier'] = pd.qcut(
    df['food_insecurity'],
    q=3,
    labels=['Low Risk', 'Moderate Risk', 'High Risk'],
)

tier_summary = df.groupby('Food_Insecurity_Tier', observed=False)[
    'diabetes_prevalence'
].agg(['mean', 'median', 'std', 'count'])
print('--- Risk Tier Breakdown ---')
print(tier_summary)

# Regional Breakdown by State
if 'StateAbbr' in df.columns:
  region_summary = (
      df.groupby('StateAbbr')[['food_insecurity', 'diabetes_prevalence']]
      .mean()
      .sort_values(by='diabetes_prevalence', ascending=False)
  )
  print('\n--- Top 5 Highest Burden States ---')
  print(region_summary.head(5))

# Correlation & Regr
corr = df['food_insecurity'].corr(df['diabetes_prevalence'])
print(f'\nPearson Correlation (r): {corr:.4f}')
print(f'Coefficient of Determination (R²): {corr**2:.4f}')
```
