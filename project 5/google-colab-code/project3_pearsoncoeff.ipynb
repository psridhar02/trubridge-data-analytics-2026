```python
import pandas as pd
from scipy.stats import pearsonr

df = pd.read_csv("/content/Sept26_FINAL_sridhar_analysis_ready_dataset.csv")
print(df.shape)
```

```python
# Pearson correlation
r, p_value = pearsonr(df['food_insecurity'], df['diabetes_prevalence'])
print(f"r= {r:.3f}, p= {p_value:.4f}")
```

```python
# Interpret Strength and Direction
abs_r = abs(r)
if abs_r < 0.1:
  strength = 'very weak'
elif abs_r < 0.3:
  strength = 'weak'
elif abs_r < 0.5:
  strength = 'moderate'
else:
  strength = 'strong'

direction = 'positive' if r > 0 else 'negative'
```

```python
# Results
n = df[['food_insecurity', 'diabetes_prevalence']].dropna().shape[0]

print(f'Pearson correlation: r = {r:.3f}, p = {p_value:.4f}, n = {n}')
print(f'This is a {strength} {direction} correlation.')
```
