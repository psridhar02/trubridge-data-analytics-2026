```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/Sept26_FINAL_sridhar_analysis_ready_dataset.csv")
```

```python
print("Stats:")
print(df[['food_insecurity', 'diabetes_prevalence']].describe())
```

```python
print("Missing Value Audit:")
print(df[['food_insecurity', 'diabetes_prevalence']].isnull().sum())
```

```python
# Comparison scatter plot
plt.figure(figsize=(9,6))
plt.scatter(df['food_insecurity'], df['diabetes_prevalence'], alpha=0.5, color='steelblue', edgecolors='none')

plt.xlabel('Food Insecurity Rate (%)')
plt.ylabel('Diagnosed Diabetes Prevalence (%)')
plt.title('County-Level Food Insecurity vs. Diagnosed Diabetes Prevalence in U.S. Counties', fontweight='bold')
plt.grid(True, linestyle='--', alpha = 0.5)
plt.tight_layout()
plt.show()
```
