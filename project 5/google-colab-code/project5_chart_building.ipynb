```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Make sure non-null data for 2 research variables
df = pd.read_csv("/content/Sept26_FINAL_sridhar_analysis_ready_dataset.csv")
chart_df = df.dropna(subset=['food_insecurity', 'diabetes_prevalence'])
```

```python
sns.set_theme(style='white') # white style

plt.figure(figsize=(9,6), dpi=150)
sns.regplot(
    data=chart_df,
    x='food_insecurity',
    y='diabetes_prevalence',
    scatter_kws={'alpha': 0.35, 'color': '#1f77b4', 's': 25},
    line_kws={'color': '#d62728', 'linewidth': 2.5},
)

plt.title(
    'Counties with Higher Food Insecurity Rates Have Higher Diabetes'
    ' Prevalence',
    fontsize=13,
    pad=15,
    fontweight='bold',
    loc='left'
)
plt.xlabel('County Food Insecurity Rate (%)', fontsize=11, labelpad=8)
plt.ylabel('Diagnosed Adult Diabetes Prevalence (%)', fontsize=11, labelpad=8)
sns.despine()
plt.savefig('food_insecurity_vs_diabetes_chart.png', dpi=150, bbox_inches='tight')
plt.show()

# 7. Download PNG
files.download('food_insecurity_vs_diabetes_chart.png')
```
