```python
import pandas as pd

df = pd.read_csv("/content/PLACES__Local_Data_for_Better_Health,_County_Data,_2025_release_20260906.csv")

# Cleaning data to only show rows with Crude prevalence
crude_df = df[df['Data_Value_Type'] == 'Crude prevalence']

# Pivot so each row is a county (location id), and columns are measure ID
df = crude_df.pivot(index=['LocationID', 'StateAbbr', 'LocationName'], columns='MeasureId', values='Data_Value').reset_index()
```

```python
# STEP 1: Rename the new columns
df = df.rename(columns={
    'FOODINSECU' : 'food_insecurity',
    'DIABETES' : 'diabetes_prevalence'
})
```

```python
# columns I want to keep in the new file
columns_to_keep = ['LocationID', 'StateAbbr', 'LocationName', 'diabetes_prevalence', 'food_insecurity']
df = df[columns_to_keep]

print("Step 1 complete - filtered columns in DataFrame")
print(df.columns.tolist())
```

```python
# STEP 2: Check and correct data types
print("Step 2 complete - updated Data Types:")
print(df.dtypes)
```

```python
# Step 3: Drop duplicated
print("Shape before dropping duplicate rows:", df.shape)
df = df.drop_duplicates()
print('Step 3 complete - shape after dropping duplicate rows:', df.shape)
```

```python
# Step 4 dropping missing values across BOTH research variables
df = df.dropna(subset=['diabetes_prevalence', 'food_insecurity'])

print('\nPost cleaning missing values counts:')
print(df.isnull().sum())
print('\nFinal dataset shaped (N = 2,300 counties):')
print(df.shape)
# Step 5: Confirm the vars are clean
print("Step 5 complete - clean variables summary")
print("Clean Summary Stats:")
print(df.describe())
```

```python
# Final validation check before saving a new csv file
print("test #1: data types")
print(df.dtypes)

print("test #2: check for remaining missing values")
print(df.isnull().sum())

print("test #3: final shape & row count")
print(df.shape)
```
